```python
from transformers import BertTokenizer, GPT2Tokenizer

# Input text
text = "Hello world"

# Load the tokenizers
bert_tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
gpt_tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

# Perform tokenization
bert_tokens = bert_tokenizer.tokenize(text)
gpt_tokens = gpt_tokenizer.tokenize(text)

# Convert tokens into token IDs
bert_ids = bert_tokenizer.convert_tokens_to_ids(bert_tokens)
gpt_ids = gpt_tokenizer.convert_tokens_to_ids(gpt_tokens)

# Display BERT results
print("BERT TOKENIZER")
print("Tokens:", bert_tokens)
print("Token IDs:", bert_ids)
print("Encoded IDs:", bert_tokenizer.encode(text))

# Display GPT-2 results
print("\nGPT-2 TOKENIZER")
print("Tokens:", gpt_tokens)
print("Token IDs:", gpt_ids)
print("Encoded IDs:", gpt_tokenizer.encode(text))

# Compare the number of content tokens
print("\nCOMPARISON")
print("BERT token count:", len(bert_tokens))
print("GPT-2 token count:", len(gpt_tokens))
```
